```python
import pandas as pd

import numpy as np

import matplotlib.pyplot as plt

import seaborn as sns

from seaborn import heatmap
```

```python
%matplotlib inline
import matplotlib.pyplot as plt

plt.style.use("ggplot")

pd.set_option("display.width", 5000)

plt.rcParams["figure.figsize"] = (10, 7)
```

```python
# Reading the Toronto weather data as a Pandas DataFrame 
Toronto_Weather = pd.read_csv("weather_2012.csv")
```

```python
# Renaming the columns for better analysis
Toronto_Weather = Toronto_Weather.rename(columns = {
    "Date/Time": "date_time", "temp (C)": "temp", "Dew Point Temp (C)": "dew_point", "Rel Hum (%)": "rel_hum", "Wind Spd (km/h)": "wind_spd", "Visibility (kn)": "visibility", "Stn Press (kPa)": "stn_press", "Weather": "weather"
})
```

```python
# target "date_time" inside Toronto_Weather, convert the text-based dates into official timestamp objects (datetime64)
# index extracts days from 1st - 31st and stores the numbers into a variable (gets the column, uses the Pandas Datetime accessor to extract)
# columns extracts months from 1 (January) to 12 (December) and stores the numbers into a variable

Toronto_Weather["date_time"] = pd.to_datetime(Toronto_Weather["date_time"]) 

Heatmap = Toronto_Weather.pivot_table(
    index=Toronto_Weather["date_time"].dt.day,
    columns=Toronto_Weather["date_time"].dt.month,
    values="Temp (C)",
    aggfunc="mean"
)
```

```python
# cmap = color map which converts numbers into colours to represent numeric data visually. 
# annot = False hides numeric values inside the heatmap cells for better visualisation of this large dataset 

sns.heatmap(Heatmap, cmap="coolwarm", annot=False)
plt.title("Average Monthly Toronto Temperature in 2012")
plt.xlabel("Month (Jan - Dec)")
plt.ylabel("Day (1st - 31st)")
plt.show()
```

## Insights
- January and February were generally cold months where temperatures were largely below freezing point, where the highest was 2.7 degrees on 22nd February.

- Temperatures started rising in April with lowest being 2.3 on 1st APril and highest with 14 - 17 April the hottest at above 10 degrees. Temperatures for the rest of the month stayed below 10 degrees. Temperatures in May started rising, remaining above 10 degrees all month long except 1st May where it averaged at 9.4 degrees.

- Temperatures between June to September oscillated between 15 - 28 degrees with exception of 2nd and 4th June where it was 14 and 13 degrees respectively.

- October marked the start of temperatures dropping to about 12 degrees on average but going as low as 4.7 degrees. It dropped to frezing point in November and December.

- This data presents a typical temperate climate which experience all the four seasons where November to February is winter, March to May is spring, which ushers in summer between June to August with autumn coming up in September/October.
